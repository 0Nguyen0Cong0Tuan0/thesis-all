```python
import os
import getpass
from huggingface_hub import snapshot_download

HF_REPO_ID = "DiBiay/video-demo-thesis"
GDRIVE_SAVE_DIR = "/content/drive/MyDrive/demo_video_counter"

# Đọc HF_TOKEN từ môi trường, Colab Userdata hoặc yêu cầu nhập an toàn
HF_TOKEN = os.environ.get("HF_TOKEN", "")
if not HF_TOKEN:
    try:
        from google.colab import userdata
        HF_TOKEN = userdata.get("HF_TOKEN")
    except Exception:
        pass

if not HF_TOKEN:
    HF_TOKEN = getpass.getpass("🔑 Nhập HF_TOKEN của bạn (hoặc nhấn Enter nếu Repo public): ")

print(f"📥 Đang tải toàn bộ Dataset Repo '{HF_REPO_ID}' từ HuggingFace về Google Drive...")
downloaded_path = snapshot_download(
    repo_id=HF_REPO_ID,
    repo_type="dataset",
    local_dir=GDRIVE_SAVE_DIR,
    token=HF_TOKEN if HF_TOKEN else None,
    max_workers=8
)

print(f"🎉 ĐÃ TẢI THÀNH CÔNG TOÀN BỘ DỮ LIỆU BÌNH THƯỜNG VỀ GOOGLE DRIVE!")
print(f"📂 Đường dẫn lưu trữ trên Google Drive: {GDRIVE_SAVE_DIR}")
```

```python
import os

GDRIVE_SAVE_DIR = "/content/drive/MyDrive/demo_video_counter"
print(f"📋 Danh sách tệp và thư mục đã tải về tại {GDRIVE_SAVE_DIR}:")
print("=" * 60)
for root, dirs, files in os.walk(GDRIVE_SAVE_DIR):
    level = root.replace(GDRIVE_SAVE_DIR, '').count(os.sep)
    indent = ' ' * 4 * level
    rel_path = os.path.relpath(root, GDRIVE_SAVE_DIR)
    if rel_path == ".":
        print(f"📁 demo_video_counter/")
    else:
        print(f"{indent}📁 {os.path.basename(root)}/")
    subindent = ' ' * 4 * (level + 1)
    for f in files[:10]:
        print(f"{subindent}📄 {f}")
    if len(files) > 10:
        print(f"{subindent}... và thêm {len(files) - 10} tệp khác")
print("=" * 60)
```
